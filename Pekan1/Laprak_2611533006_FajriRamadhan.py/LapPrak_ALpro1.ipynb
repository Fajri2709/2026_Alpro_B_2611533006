## 1.1 Latar Belakang
Perkembangan teknologi informasi membuat kemampuan pemrograman menjadi salah satu keterampilan yang penting untuk dipelajari. Pemrograman dapat digunakan untuk membuat aplikasi, mengolah data, mengembangkan sistem informasi, hingga menyelesaikan berbagai permasalahan secara otomatis.

Salah satu bahasa pemrograman yang banyak digunakan adalah Python. Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks relatif sederhana sehingga sesuai untuk pemula. Python juga memiliki banyak pustaka yang dapat digunakan dalam berbagai bidang, seperti pengembangan aplikasi, kecerdasan buatan, analisis data, dan otomasi.

Selain editor kode, Git dan GitHub juga berperan penting dalam pengembangan perangkat lunak. Git digunakan untuk mengelola perubahan pada kode secara lokal, sedangkan GitHub merupakan platform yang dapat digunakan untuk menyimpan dan mengelola repository secara daring. Dengan menghubungkan VS Code dengan GitHub, pengguna dapat mengunggah kode ke repository dan melakukan pengelolaan proyek dengan lebih mudah.

**Soal : Membuat sebuah file program Python bebas dengan nama tugas1_nim.py.**

```python
print("=== Program Persegi Panjang ===")

panjang = float(input("Masukkan panjang: "))
lebar = float(input("Masukkan lebar: "))

luas = panjang * lebar
keliling = 2 * (panjang + lebar)

print("\nHasil Perhitungan")
print("Luas      :", luas)
print("Keliling  :", keliling)
```
